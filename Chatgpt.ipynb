```python
while True:
    user = input("enter the input :").lower()
    response = {
        "hello" : "hi there",
        "how are you" : "i am doing fine",
        "your name" : "i am Minichat"
    }
    if user in response:
        print(response[user])
    elif user == "bye":
        print("Goodbye")
        break 
    else:
        print("i don't understand")
    
```
